pytorchを使用して、重みWを特異値分解してみる

```python
import torch
import torch.nn as nn

# ============================================================
# 目的: Linear 層の重み行列 W を SVD 分解する
# ============================================================
# 数式: W ≈ U @ diag(S) @ Vh
#   U  : 左特異ベクトル（出力側の方向）
#   S  : 特異値（重要度。大きいほど W の情報量が多い）
#   Vh : 右特異ベクトル（入力側の方向、V の転置）

# MNIST の MLP 1層目を想定: 784次元 → 512次元
layer = nn.Linear(784, 512)

# PyTorch の Linear.weight は (out_features, in_features) = (512, 784)
# 計算式: y = x @ W.T + b  （x: バッチ×784, y: バッチ×512）
print("weight", layer.weight.shape)

W = layer.weight.data
U, S, Vh = torch.linalg.svd(W, full_matrices=False)

# full_matrices=False → thin SVD（必要な成分だけ返す）
# W が (512, 784) なので特異値は min(512, 784) = 512 個
print("W ≈ U @ diag(S) @ Vh")
print("U.shape ", U.shape)   # (512, 512) 左特異ベクトル
print("S.shape ", S.shape)   # (512,)     特異値ベクトル（Σ の対角成分）
print("Vh.shape", Vh.shape)  # (512, 784) 右特異ベクトル
print("S (先頭10個)", S[:10])  # 降順。先頭ほど重要
```
